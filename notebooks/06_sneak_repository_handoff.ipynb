Verifies the declared Git snapshot, actual bytes, Python/Node roots and meaningful unit checks. It does not validate application integration or semantic truth.


```python
from pathlib import Path
import subprocess, json
ROOT = Path.cwd()
if not (ROOT / 'scripts/custody_snapshot.py').exists(): ROOT = ROOT.parent
assert (ROOT / 'scripts/custody_snapshot.py').exists()
def run(args):
    p = subprocess.run(args, cwd=ROOT, text=True, capture_output=True)
    print(p.stdout); print(p.stderr)
    assert p.returncode == 0, (args, p.returncode)
    return p.stdout
```
